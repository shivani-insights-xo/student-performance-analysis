# Student Performance Analysis

Objective:

The objective of this project is to analyze the Student Performance dataset using Python. The dataset was cleaned, analyzed, and visualized to understand factors affecting student grades.

# Importing Libraries and loading the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("student-mat.csv", sep=";")
```

Explanation:

Pandas is used for data loading and analysis.
Matplotlib is used for data visualization.
The dataset was loaded imto a pandas DataFrame for further analysis.


# Exploring the dataset

```python
df.head()
df.info()
df.shape
```

Explanation:

head() displays the first five rows.

info() provides column information and data types.

shape shows the number of rows and columns.

# Data Cleaning

```python
df.isnull().sum()
df.duplicated().sum()
```

Explanation:

The dataset was checked for:

Missing values

Duplicate records

Any duplicate rows were removed to improve data quality.

Analysis 1: Average Final Grade

```python
average_grade = df['G3'].mean()
print("\nAverage Final Grade (G3):", round(average_grade, 2))
```

Observation:

This calculates the average final grade (G3) of all students.

Analysis 2: Students Scoring Above 15

```python
len(df[df['G3'] > 15])
```

Observation:

This identifies the number of high-performing students score above than 15.

Analysis 3: Correlation Between Study Time and Grades

```python
df['studytime'].corr(df['G3'])
```

Observation:

The correlation value indicates whether study time influences final grades.

Analysis 4: Performance by Gender

```python
df.groupby('sex')['G3'].mean()
```

Observation:

This compares the average grades of male and female students.

Visualization 1: Distribution of Final Grades

```python
plt.hist(df['G3'])
plt.title("Distribution of Final Grades")
plt.xlabel("Grades")
plt.ylabel("Frequency")
plt.show()
```

Observation

The histogram shows how student grades are distributed.

Visualization 2: Study Time vs Final Grade

```python
plt.scatter(df['studytime'], df['G3'])
plt.title("Study Time vs Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")
plt.show()
```

Observation

The scatter plot helps visualize the relationship between study time and academic performance.

Visualization 3: Average Grade by Gender

```python
gender_avg = df.groupby('sex')['G3'].mean()

plt.figure(figsize=(6, 5))
gender_avg.plot(kind='bar')
plt.title("Average Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Grade")
plt.show()

plt.show()
```

Observation

The bar chart compares average performance between genders.
